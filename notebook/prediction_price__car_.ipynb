## outlier

```python
data_numeric = df.select_dtypes(exclude='object')
for col in data_numeric.columns:
  q1=df[col].quantile(0.25)
  q3=df[col].quantile(0.75)
  iqr=q3-q1
  lower_bound=q1-1.5*iqr
  upper_bound=q3+1.5*iqr
  outliers=df[(df[col]<lower_bound)|(df[col]>upper_bound)]
  df.drop(outliers.index,inplace=True)
```

```python
from sklearn.linear_model import LinearRegression,Lasso,Ridge,ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.ensemble import BaggingRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.ensemble import VotingRegressor
from sklearn.ensemble import StackingRegressor
from sklearn.tree import ExtraTreeRegressor
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.ensemble import ExtraTreesRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
```

```python
from math import gamma
models={
    #"LinearRegression":LinearRegression(),
    #"Lasso":Lasso(),
    #"Ridge":Ridge(),
    #"ElasticNet":ElasticNet(),
    #"knn":KNeighborsRegressor(n_neighbors=20,metric="manhattan"),
    #"DecisionTreeRegressor":DecisionTreeRegressor(max_depth=15),
    #"RandomForestRegressor":RandomForestRegressor(n_estimators=100,max_depth=15),
    #"SVR":SVR(C=100,gamma=1),
   # "GradientBoostingRegressor":GradientBoostingRegressor(n_estimators=100,learning_rate=0.1,max_depth=20),
   # "BaggingRegressor":BaggingRegressor(n_estimators=10),
   # "AdaBoostRegressor":AdaBoostRegressor(n_estimators=100,learning_rate=0.1),
   # "VotingRegressor":VotingRegressor(estimators=[("lr",LinearRegression()),("knn",KNeighborsRegressor())]),
    "StackingRegressor":StackingRegressor(estimators=[("lr",LinearRegression()),("knn",KNeighborsRegressor()),("dt",DecisionTreeRegressor()),("ba",BaggingRegressor()),("et",ExtraTreeRegressor())]),
    #"ExtraTreeRegressor":ExtraTreeRegressor(),
    #"HistGradientBoostingRegressor":HistGradientBoostingRegressor(max_iter=100,max_depth=3),
    "ExtraTreesRegressor":ExtraTreesRegressor(n_estimators=150,max_depth=35),
    "XGBRegressor":XGBRegressor(n_estimators=100,learning_rate=0.1,max_depth=50)
}
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
accuracy=[]
for model_name , model in models.items():
    model.fit(x_train,y_train)
    y_pre_train=model.predict(x_train)
    y_pre_test=model.predict(x_test)
    accuracy.append({
        "model":model_name,
        "MSE_train":mean_squared_error(y_train,y_pre_train),
        "MSE_test":mean_squared_error(y_test,y_pre_test),
        "MAE_train":mean_absolute_error(y_train,y_pre_train),
        "MAE_test":mean_absolute_error(y_test,y_pre_test),
        "R2_train":r2_score(y_train,y_pre_train),
        "R2_test":r2_score(y_test,y_pre_test)
    })
```

```python
pd.DataFrame(accuracy).sort_values(by="R2_test",ascending=False)
```

```python
fig, ax = plt.subplots(figsize=(20,5))
accuracy_df = pd.DataFrame(accuracy).sort_values(by="R2_test",ascending=False)
ax.plot(accuracy_df.model, accuracy_df.R2_test, label="R2 Score")
ax.set_xlabel('Model')
ax.set_ylabel('R2 Score')
ax.set_title('R2 Score of Models')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

## Hyperparameter Tuning with Randomized Search

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint

param_dist_etr = {
    'n_estimators': randint(50, 300),
    'max_depth': randint(10, 50),
    'min_samples_split': randint(2, 20),
    'min_samples_leaf': randint(1, 10),
    'bootstrap': [True, False]
}

etr = ExtraTreesRegressor(random_state=42)

random_search_etr = RandomizedSearchCV(
    estimator=etr,
    param_distributions=param_dist_etr,
    n_iter=50,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

random_search_etr.fit(x_train, y_train)

print("Best parameters for ExtraTreesRegressor:", random_search_etr.best_params_)
print("Best R2 score on validation set:", random_search_etr.best_score_)
```

```python
# Evaluate the best model on the test set
best_etr_model = random_search_etr.best_estimator_
y_pred_test_etr = best_etr_model.predict(x_test)
r2_test_etr = r2_score(y_test, y_pred_test_etr)
mse_test_etr = mean_squared_error(y_test, y_pred_test_etr)
mae_test_etr = mean_absolute_error(y_test, y_pred_test_etr)

print(f"Test R2 Score with best ExtraTreesRegressor: {r2_test_etr:.4f}")
print(f"Test MSE with best ExtraTreesRegressor: {mse_test_etr:.2f}")
print(f"Test MAE with best ExtraTreesRegressor: {mae_test_etr:.2f}")

accuracy.append({
    "model": "ExtraTreesRegressor_RandomSearch",
    "MSE_train": mean_squared_error(y_train, best_etr_model.predict(x_train)),
    "MSE_test": mse_test_etr,
    "MAE_train": mean_absolute_error(y_train, best_etr_model.predict(x_train)),
    "MAE_test": mae_test_etr,
    "R2_train": r2_score(y_train, best_etr_model.predict(x_train)),
    "R2_test": r2_test_etr
})

pd.DataFrame(accuracy).sort_values(by="R2_test",ascending=False)
```
